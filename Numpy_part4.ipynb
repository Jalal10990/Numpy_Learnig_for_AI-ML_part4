```python
# MODULE 06 — NumPy Array Operations & Aggregations
# The core mental model: Two families of operations
# Imagine that you have a dataset containing the marks of 1,000 students. Sometimes you want to change every value. Other times, you want to calculate one summary, such as the average mark.
# Family 1: Element-wise operations --- Calculate matching elements individually.
# For example, [10, 20, 30] + [2, 4, 5] produces [12, 24, 35].

# Family 2: Reductions and aggregations
# For example, [10, 20, 30] + [2, 4, 5] produces [12, 24, 35].
# For example, np.mean([10, 20, 30]) returns the single number 20.0

# @ (matrix multiplication) is a linear-algebra operation, not an ordinary element-wise operation or a reduction. It combines elements through dot products and may change the shape of the array.

# Part 1 — Element-wise operations
# 6.1 Positional arithmetic and scalar math
import numpy as np
a= np.array([10,20,30])
b = np.array([2,4,6])

# Think of them as two spreadsheets placed side by side. Each value in a is matched with the value at the same position in b.
# Operator	Meaning	Example
# +	Addition	a + b
# -	Subtraction	a - b
# *	Element-wise multiplication	a * b
# /	Element-wise division	a / b
# **	Element-wise exponentiation	a ** 2

print(a+b)
print(a - b)
print(a * b)
print(a / b)
# NumPy's ordinary / operation produces floating-point results, even when the division happens to be exact.
print(a ** 2)
```

```python
# Scalar operations: One number applied to every element
# A scalar is a single numerical value, such as 10 or 5
a = np.array([10,20,30])
print(a* 10)
print(a + 5)
print(a ** 2)

# Suppose your dataset stores temperatures in Celsius and you need Fahrenheit. You can convert the entire array without writing a loop:
celsius =np.array([0,20,30,100])
fahrenheit = celsius * 9/5 +32
print(fahrenheit)

```

```python
# Broadcasting — How NumPy handles different shapes
# What happens if the two arrays do not have exactly the same shape?
# NumPy can sometimes perform element-wise operations on differently shaped arrays by logically expanding the smaller shape to match the larger one.
A = np.array([
    [10,20,30],
    [40,50,60]
])
b = np.array([1,2,3])
print(A+b)

# NumPy does not necessarily create a physical copy of the vector. Broadcasting describes how the operation treats the shapes.
# Broadcasting rules
# Compare shapes from right to left. Two dimensions are compatible if:
# - They are equal, or
# - One of them is 1, or
# - One shape has no dimension remaining on that side.
# Shape A	Shape B	Compatible?	Result shape
# (2, 3)	(2, 3)	Yes	(2, 3)
# (2, 3)	(3,)	Yes	(2, 3)
# (2, 3)	(2, 1)	Yes	(2, 3)
# (2, 3)	(2,)	No	Error




```

```python
# The crucial distinction: * vs @
# Element-wise multiplication versus matrix multiplication
# A @ B — Matrix multiplication
# Matrix multiplication follows a different rule: multiply a row from the first matrix by a column from the second matrix, then add the products.
# A * B — Element-wise multiplication
# Each element is multiplied by the element in the same position.
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("Element-wise multiplication:")
print(A * B)

print("Matrix multiplication:")
print(A @ B)

```

```python
# Matrix multiplication and shape compatibility
# Suppose:

# A\text{ has shape }(m,n)

# and
# B\text{ has shape }(n,p)

# The inner dimensions must match. The resulting matrix has shape:
# (m,n)@(n,p)\rightarrow(m,p)
import numpy as np
A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
B = np.array([
    [10, 20],
    [30, 40],
    [50, 60]
])
C = A @ B

print(C)
print(C.shape)


```

```python
import numpy as np

X = np.array([
    [2, 80],
    [5, 90]
], dtype=float)

W = np.array([
    [0.5, 1.0, -1.0],
    [0.1, 0.2, 0.5]
])

b = np.array([1.0, 2.0, 0.5])

Z = X @ W + b

print(Z)
print("Output shape:", Z.shape)
```

```python
# Statistical reductions and aggregations
# What is a reduction?
 # an element_wise operation usually produves an array with same shape as its input.A reduction summarize multiply values into fewer values.
scores = np.array([60, 70, 80, 90])
# You might want to know:
# - The total score.
# - The average score.
# - The minimum and maximum.
# - How spread out the scores are.
# - The position of the highest score
# Function	Purpose
# np.sum()	Add values
# np.mean()	Calculate the average
# np.min()	Find the minimum
# np.max()	Find the maximum
# np.std()	Calculate standard deviation
# np.var()	Calculate variance
# np.argmax()	Return the index of the maximum
# np.argmin()	Return the index of the minimum
print(np.mean(scores))
print(np.std(scores))
print(np.var(scores))
# Variance and standard deviation describe how spread out values are around the mean
# The mean is 75. The deviations from the mean are:
# -15, -5, 5, 15
# Squaring the deviations gives:
# 225, 25, 25, 225
# NumPy's default population variance is the average of those squared deviations:
# \[
# \operatorname{var}(x)
# =\frac{225+25+25+225}{4}
# =125
# \]

# Standard deviation is the square root of the variance:
# \[
# \operatorname{std}(x)=\sqrt{125}\approx11.18
# \]
```

```python
# The axis parameter
# the axis parameter tells NumpPy which dimension to reduce.
# rows represent students and columns represent subjects
marks = np.array([
    [80,70,90],
    [60,85,75],
    [95,90,85]
])
# A. axis=0 — Reduce down the rows
print(np.mean(marks, axis=0))
# This calculates one average per column, giving the average mark in each subject.
# The three calculations are:
# \[
# \begin{aligned}
# \text{Subject 1}&=(80+60+95)/3\\
# \text{Subject 2}&=(70+85+90)/3\\
# \text{Subject 3}&=(90+75+85)/3
# \end{aligned}
# \]
# The output shape is (3,).


# B. axis=1 — Reduce across the columns
print(np.mean(marks, axis=1))


# Expression	Meaning	Output shape
# np.sum(marks)	Sum all nine values	Scalar
# np.mean(marks, axis=0)	Average for each subject	(3,)
# np.mean(marks, axis=1)	Average for each student	(3,)
# np.sum(marks, axis=0)	Total for each subject	(3,)
# np.sum(marks, axis=1)	Total for each student	(3,)

# axis=0 removes the row dimension by combining values down the rows. axis=1 removes the column dimension by combining values across each row.
```

```python
# # Part 5 — Loss functions: Connecting NumPy to learning
# 6.8 What is a loss function? 

# A Loss fnction converts the difference between predictions and actual answer into a numerical measure of error
# During training, a learning algorithm adjusts the model's parameters to reduce the loss.
# A lower loss usually means predictions are closer to the correct answers according to the chosen loss function. However, the loss alone does not always tell us how well a model will perform on new data.

# Mean Squared Error (MSE)
# Mean Squared Error is commonly used for regression tasks, where the model predicts numerical values such as temperatures, prices or scores.
# Its formula is:
# \[
# \boxed{\operatorname{MSE}
# =\frac{1}{n}\sum_{i=1}^{n}(y_i-\hat y_i)^2}
# \]

# Where:
# - \(y_i\) = actual value.
# - \(\hat y_i\) = predicted value.
# - \(n\) = number of predictions.
# - Squaring makes negative and positive errors contribute positively.

actual = np.array([100,200,300])
predicted = np.array([110,190,280])

# Step 1 — Calculate the errors
# y-\hat y=[-10,10,20]
# Step 2 — Square each error
# [-10,10,20]^2=[100,100,400]
# Step 3 — Calculate their average
# {MSE}=\frac{100+100+400}{3}=200

errors = actual - predicted
squared_errors = errors **2
mse = np.mean(squared_errors)

print("Errors:", errors)
print("Squared errors:", squared_errors)
print("MSE:", mse)

# //alternate mse = np.mean((actual - predicted) ** 2)


# Mean Absolute Error (MAE)
# Another common regression loss is Mean Absolute Error.
# \[
# \boxed{\operatorname{MAE}
# =\frac{1}{n}\sum_{i=1}^{n}|y_i-\hat y_i|}
# \]

# Instead of squaring the errors, MAE takes their absolute values.
mae = np.mean(np.abs(actual - predicted))

print("MAE:", mae)

```

```python
# Part 2: Statistical Reductions & Aggregations
# Aggregations: Many Numbers → One Summary
# What problem does aggregation solve?
# Imagine you are a teacher and have the marks of 1,000 students. Looking at every student's marks individually would take a lot of time. Instead, you might want to know:
# - What is the total of all marks?
# - What is the average score?
# - How spread out are the scores?
# - What is the highest or lowest score?
# This is the purpose of aggregation, also called reduction.
# An aggregation takes multiple values and produces a summary value
# Understand each aggregation
# NumPy function	Meaning	Question it answers
# np.sum(a)	Adds values	What is the total?
# np.mean(a)	Calculates average	What is the center?
# np.var(a)	Measures squared spread	How much do values vary?
# np.std(a)	Measures spread in original units	How far do values typically spread from the mean?
# np.min(a)	Finds the smallest value	What is the lowest value?
# np.max(a)	Finds the largest value	What is the highest value?
import numpy as np

scores = np.array([45, 88, 92, 55, 78])

print("Total Sum :", np.sum(scores))
print("Mean      :", np.mean(scores))
print("Variance  :", np.var(scores))
print("Std Dev   :", np.std(scores))
print("Minimum   :", np.min(scores))
print("Maximum   :", np.max(scores))
```

```python
# The golden distinction: max() vs. argmax()
# This distinction is especially important in machine learning.
# Remember this rule:
# - np.max() tells you the value.
# - np.argmax() tells you the index position of that value.
import numpy as np

arr = np.array([10, 50, 30])

print("Maximum value:", np.max(arr))
print("Maximum index:", np.argmax(arr))

print("Minimum value:", np.min(arr))
print("Minimum index:", np.argmin(arr))
```
